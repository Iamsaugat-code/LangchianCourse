
### Tool 
A tool in LangChain is a function or external service that an AI model can use to perform a specific task instead of only generating text. For example, a tool can be used to search the web, calculate numbers, access a database, or get weather information. The AI decides when a tool is needed and uses it to get the required result.

Example: A calculator tool can help the AI calculate 25 × 10 and return the answer.

```python
import langchain

langchain.__version__
```

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python

#  creating model

import os 

from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:openai/gpt-oss-20b")

respone = model.invoke("what is ai ?")

respone.content
```

```python

# creating toold

from langchain.tools import  tool

@tool
def weather(location:str)->str:
    """get weather at location"""
    return f"it is sunny in {location}"

model_tools = model.bind_tools([weather])

respone = model_tools.invoke("what is the weather in pokhara ?")

print(respone)

for i in respone.tool_calls:
    print(f"tool : {i['name']}")
    print(f"args : {i['args']}")
```

```python

from langchain.agents import  create_agent

agent = create_agent(
    model = "groq:openai/gpt-oss-20b",
    tools = [weather],
    system_prompt = "this ai asstiant"
)
res = agent.invoke({"messages":"what is the weather of pokhara ?"})
res["messages"][-2].content
```


### Tool execution 

```python

messages = [{'role':'user','content':'what is the weather in jhapa'}]

ai_msg = model_tools.invoke(messages)

messages.append(ai_msg)

for i in ai_msg.tool_calls:
    result = weather.invoke(i)
    messages.append(result)

final_respone = model_tools.invoke(messages)
print(final_respone.text)
messages
```

### Messages 

In LangChain, a message represents communication between the user, AI, or system. It mainly contains role, content, and metadata. Role tells who sent the message, such as system, human, or ai. Content contains the actual text or information of the message. Metadata stores additional information about the message, such as message ID, model name, timestamp, or other useful details. For example, a human message may have the role human, content "What is Python?", and metadata containing information about the request.


### Message Prompt
A message prompt is a structured prompt used in LangChain where the conversation is divided into different message types based on the sender or purpose. The main types are:

Human: Message or question given by the user.
System: Instructions that define the AI's behavior or role.
AI: Response generated by the AI model.
Tool: Message containing the result or output returned by a tool used by the AI.

Example:<br>
System: You are a helpful assistant.<br>
Human: What is Python?<br>
AI: Python is a programming language.<br>
Tool: Weather data = 25°C.


#### System Message

A System message provides instructions or guidelines that define how the AI should behave and respond. It usually sets the AI's role, personality, or rules before the conversation starts.
Example: “You are a helpful programming teacher.


#### Human  message

A Human message represents the input, question, or instruction provided by the user to the AI model. It contains what the user wants the AI to understand or perform.
Example: “Explain what Python is.”

#### AI Messages

An AI message represents the response generated by the AI model after processing the user's input and system instructions. It contains the answer or information provided by the AI.
Example: “Python is a high-level programming language.”

#### Tool Messages

A Tool message contains the result or output returned by an external tool after the AI requests it. It is used when the AI needs additional information, such as weather, database results, calculations, or API responses.
Example: “Temperature in Kathmandu: 25°C.”

```python

### tools call

from langchain.messages import ToolMessage
from langchain.messages import  AIMessage

ai_mss = AIMessage(
    content = [],
    tool_calls = [{
        "name":"weather",
        "args":{"location":"Kathmandu"},
        "id" : "123"
    }]
)

weather_s = "sunny, 42c"

tool_mss = ToolMessage(
    content = weather_s,
    tool_call_id = "123"
)

mss = [
    HumanMessage("What is the weather of Kathmandu ? "),
    ai_mss,
    tool_mss
]

res = model.invoke(mss)
print(res.content)

```
